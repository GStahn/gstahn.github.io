```python
import os
import pandas as pd

os.chdir('GitHub/myside/assets/downloads') # change to your direction

df = pd.read_csv("./data_ofr_varmetric_adjusted.csv", sep=";")

df.head()
```

```python
### Create shares ###

df['Total'] = df['Qualifying Hedge Funds: no risk metric reported (count)'] + df['Qualifying Hedge Funds: calculates alternative to VaR (count)'] + df['Qualifying Hedge Funds: calculates VaR (count)']

df['Share_nometric'] = (df['Qualifying Hedge Funds: no risk metric reported (count)']/df['Total'])*100
df['Share_alt'] = (df['Qualifying Hedge Funds: calculates alternative to VaR (count)']/df['Total'])*100
df['Share_var'] = (df['Qualifying Hedge Funds: calculates VaR (count)']/df['Total'])*100

df["date"] = pd.to_datetime(df["date"], errors="coerce") 
df["quarter"] = df['date'].dt.to_period("Q")

df.head()
```

```python
### Create plot ###

import matplotlib.pyplot as plt
import matplotlib.dates as mdates

# make sure data is sorted in time
df = df.sort_values("date")

fig, ax = plt.subplots()

ax.plot(df["date"], df["Share_nometric"],
        label="No risk metric reported",
        linewidth=2)

ax.plot(df["date"], df["Share_alt"],
        label="Alternative to VaR",
        linewidth=2)

ax.plot(df["date"], df["Share_var"],
        label="VaR",
        linewidth=2)

ax.set_xlabel("Year")
ax.set_ylabel("Share of qualifying hedge funds (%)")
ax.set_ylim(0, 100)

ax.xaxis.set_major_locator(mdates.YearLocator())
ax.xaxis.set_major_formatter(mdates.DateFormatter("%Y"))

ax.grid(axis="y")   # subtle grid, y-axis only
ax.legend()

plt.tight_layout()
plt.savefig("hedge_fund_risk_metric_shares.svg", bbox_inches="tight")

### Save plot ###

plt.savefig(
    "hedge_fund_risk_metric_shares.png",
    dpi=300,
    bbox_inches="tight",
    facecolor="white"
)
plt.show()
```
